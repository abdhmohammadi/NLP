```python
expression = 'We can calculate simple math operations. Sum of 5 + 6 is equal to 11 and 5 - 6 = -1. Mathematics is lovely. contact with us email@sample.com'
```


### **Write regular expressions for the following languages.**

1. All alphabetic characters used in the expression.

2. The set of all alphabetic strings.

3. The set of all lowercase alphabetic strings ending in a specific character.

4. The set of all alphabetic strings starts with uppercase and ending in a specific character.

5. Set of all alphabetic strings that contains a substring.

6. Set of all strings from the alphabet `ab` such that each `a` is immediately preceded by and immediately followed by a `b`.

**1. Splited all alphabetic characters:**

```python
import re

# [a-z]: lower case alphabets
# [A-Z]: upper case alphabets
# [0-9]: any digits
# |: OR operator, it makes the patern returns list of matched characters
patern = re.compile('[a-z]|[A-Z]')
match = patern.findall(expression)
print('All alphabetic characters used in the expression:')
print(match)
print()
```

**2. The set of all alphabetic strings:**

```python
patern = re.compile(r'\b[a-zA-Z]+\b')
match = patern.findall(expression)
print('The set of all alphabetic strings:')
print(match)
print()
```

**3. The set of all lowercase alphabetic strings ending in a specific character:**

```python
end = 'e'
patern = re.compile(fr'\b[a-z]*{end}\b')
match = patern.findall(expression)
print(f'Set of all lowercase alphabetic strings ending in a \'{end}\':')
print(match)
print()
```

**4. Set of all alphabetic strings starts with uppercase and ending in a specific character:**

```python
end = 'm'
patern = re.compile(fr'[A-Z][a-z]*{end}+\b')
match = patern.findall(expression)
print(f'Set of all alphabetic strings starts with uppercase and ending in a \'{end}\':')
print(match)
print()
```

**5. Set of all alphabetic strings that contains a substring**

```python
substring = "ma"
pattern = re.compile(r'\b\w*' + re.escape(substring) + r'\w*\b')
matches = pattern.findall(expression)

print(f'Set of all alphabetic strings that contains a substring:\'{substring}\'.')
print(matches)
print()
```

**6. Set of all strings from the alphabet `ab` such that each `a` is immediately preceded by and immediately followed by a `b`.**

```python
text = "bb bab bbb babb ba ab a babbab"

pattern = re.compile(r'\b(?:b+|(?:bab)+)\b')

matches = pattern.findall(text)

print('Set of all strings from the alphabet ab such that each a is immediately preceded by and immediately followed by a b:')
print(matches)
print()

```
